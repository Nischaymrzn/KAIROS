# Deep learning (TabularMLP)

A CPU-sized PyTorch MLP with embeddings. Boosting usually wins on tabular data; this is the required comparison.

```
import os, sys
os.environ.setdefault('OMP_NUM_THREADS', '4')
sys.path.insert(0, os.path.abspath('..'))
from src.seeds import set_global_seed
set_global_seed(42)
from src.config import get_config
cfg = get_config()
```

```
import subprocess; subprocess.run([sys.executable,'-m','src.models.train','--model','mlp'],cwd='..',check=True)
```

**Acceptance:** the MLP trains on CPU and beats the baseline AUC.

```
import joblib,json
from src.dataset import load_processed
from src.models.registry import predict_proba
from src.models.evaluate import metrics
ds=load_processed(cfg,with_test=False); b=joblib.load('../models/mlp.joblib')
m=metrics(ds.val['MADE'],predict_proba(b,ds.val[ds.features]))
base=json.load(open('../reports/figures/baseline_metrics.json'))['val']['auc']
print('mlp AUC',round(m['auc'],4)); assert m['auc']>base
```
